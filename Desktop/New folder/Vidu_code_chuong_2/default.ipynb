```python
import json

# Hàm xử lý số phức: kiểm tra nếu obj là số phức thì trả về danh sách gồm phần thực và phần ảo
def complex_handler(obj):
    if isinstance(obj, complex):  # kiểm tra kiểu dữ liệu có phải complex không
        return [obj.real, obj.imag]  # trả về danh sách gồm phần thực và phần ảo của số phức
    raise TypeError(f"Không thể mã hóa JSON: {repr(obj)}")  # nếu không phải số phức, ném ra lỗi TypeError

# Ứng dụng hàm complex_handler để mã hóa số phức
data = {"z": 2 + 3j}

complex_obj = json.dumps(data, default=complex_handler)
# Sử dụng đối số default để chỉ định hàm xử lý tùy chỉnh complex_handler()

print(complex_obj)
```

```python
print(complex_obj, type(complex_obj))  # in ra chuỗi JSON và kiểu dữ liệu của biến complex_obj
```
