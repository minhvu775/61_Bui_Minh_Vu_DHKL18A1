```python
import json

# Đối tượng Python cần chuyển đổi
data = {
    "name": "A",
    "age": 35,
    "company": "Đất Việt"
}

print(type(data))  # data là đối tượng Python kiểu dict

# Chuyển đổi đối tượng Python thành chuỗi JSON
json_string = json.dumps(data, ensure_ascii=False)
# Hàm dumps() chuyển đối tượng Python thành chuỗi JSON,
# đối số ensure_ascii=False để giữ nguyên ký tự Unicode

print(json_string)  # In chuỗi JSON
print(type(json_string))  # json_string là chuỗi JSON kiểu str, không phải dict nữa
```

```python
import json

# Chuỗi JSON bao quanh bằng dấu nháy đơn
json_string_1 = '{"name":"A","age":35,"company":"Đất Việt"}'

# Chuỗi JSON bao quanh bằng dấu nháy kép
json_string_2 = "{\"name\":\"A\",\"age\":35,\"company\":\"Đất Việt\"}"

# Chuyển chuỗi JSON thành đối tượng Python (đối tượng JSON)
json_object_1 = json.loads(json_string_1)
json_object_2 = json.loads(json_string_2)

print(json_object_1)
print(json_object_2)
```

```python
import json
from decimal import Decimal         # cung cấp lớp Decimal để hỗ trợ cho các phép tính số học
                                    # thập phân độ chính xác cao

json_string = '{"name": "Alice", "age": 25, "balance": 1000.5, "status": "NaN"}'

data = json.loads(
    json_string,
    parse_float=Decimal,            # Chuyển đổi float thành Decimal
    parse_int=str,                  # Chuyển đổi int thành chuỗi
    parse_constant=lambda x: None  # Chuyển các giá trị đặc biệt thành None
)

print(data)
```
