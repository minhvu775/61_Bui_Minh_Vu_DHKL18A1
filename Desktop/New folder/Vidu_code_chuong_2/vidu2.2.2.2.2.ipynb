```python
import json

# Đối tượng Python (đối tượng dict)
data = {
    "Họ_và_tên": "Ng Văn A",
    "Năm_sinh": 2005,
    "Giới_tính": "Nam",
    "SĐT": "0123456789",
    "Email": "nguyenvana@uneti.edu.vn"
}

# Chuyển đổi đối tượng Python thành chuỗi JSON
json_string = json.dumps(data, ensure_ascii=False)
# Hàm dumps() chuyển đổi đối tượng Python thành chuỗi JSON,
# đối số ensure_ascii=False để giữ nguyên ký tự Unicode (tiếng Việt)

print(json_string)
print(type(json_string))
```
