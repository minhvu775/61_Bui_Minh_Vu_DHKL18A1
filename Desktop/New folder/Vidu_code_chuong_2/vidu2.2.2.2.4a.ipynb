```python
import json

# Định nghĩa một lớp đối tượng tùy chỉnh
class CustomObject:
    def __init__(self, data):
        self.data = data
    def to_dict(self):
        return {"custom_data": self.data}

# Tạo một lớp JSONEncoder tùy chỉnh để xử lý đối tượng CustomObject
class CustomObjectEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, CustomObject):
            return obj.to_dict()
        return super().default(obj)

# Tạo một đối tượng CustomObject
custom_obj = CustomObject({'key': 'value'})

# Sử dụng JSONEncoder tùy chỉnh để mã hóa đối tượng CustomObject
json_data = json.dumps(custom_obj, cls=CustomObjectEncoder)

print(json_data)
```

```python
import json

class CustomObject:                                     # Định nghĩa lớp CustomObject với hai thuộc tính name và value không được hỗ trợ trực tiếp bởi JSON
    def __init__(self, name, value):
        self.name, self.value = name, value
    def to_dict(self):                                  # Phương thức chuyển đổi đối tượng thành từ điển để có thể mã hóa JSON
        return {"__type__": "CustomObject", "name": self.name, "value": self.value} # Thêm khóa __type__ để xác định loại đối tượng và
                                                                                    # trả về từ điển với các thuộc tính name và value

class CustomObjectEncoder(json.JSONEncoder):            # Định nghĩa lớp CustomObjectEncoder kế thừa từ json.JSONEncoder để tùy chỉnh việc mã hóa JSON
    def default(self, obj):                             # Ghi đè phương thức default để xử lý đối tượng không thể mã hóa mặc định
        if isinstance(obj, CustomObject):               # Nếu obj là instance của CustomObject, gọi phương thức to_dict() để chuyển đổi thành từ điển
            return obj.to_dict()
        return super().default(obj)                     # Gọi phương thức default của lớp cha để xử lý các đối tượng khác

data = {"item": CustomObject("CT", 42)}                 # Tạo một đối tượng CustomObject và đặt nó trong từ điển data
print(json.dumps(data, cls=CustomObjectEncoder, ensure_ascii=False, indent=2)) #Sử dụng CustomObjectEncoder để mã hóa đối tượng CustomObject
                                                                                 #thành chuỗi JSON với định dạng thụt lề 2 khoảng trắng
# {
#   "item": {
#     "__type__": "CustomObject",
#     "name": "CT",
#     "value": 42
#   }
# }
```
