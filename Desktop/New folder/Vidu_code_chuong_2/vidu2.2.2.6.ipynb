```python
import json

# Đọc nội dung tệp với mã hóa UTF-8
with open('thongtin.json', 'r', encoding='utf-8') as f:
     content = f.read()

# Phân tích nội dung JSON từ chuỗi UTF-8
infor = json.loads(content)

print(infor)
print(type(infor))
```
